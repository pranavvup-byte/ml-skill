```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score

# --------------------------------------------------
# 1. Load dataset
# --------------------------------------------------

df = pd.read_csv("/content/sample_data/simple_placement_dataset.csv")

print("Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)


# --------------------------------------------------
# 2. Separate target and features
# --------------------------------------------------

y = df["Placed"]

X = df.drop(columns=["Placed"])


# --------------------------------------------------
# 3. Remove StudentID
# --------------------------------------------------

X = X.drop(columns=["StudentID"])


# --------------------------------------------------
# 4. Split dataset
# --------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("\nTraining data:", X_train.shape)
print("Testing data :", X_test.shape)


# --------------------------------------------------
# 5. Create weak learner
# --------------------------------------------------

base_model = DecisionTreeClassifier(
    max_depth=2,
    random_state=42
)


# --------------------------------------------------
# 6. Create AdaBoost model
# --------------------------------------------------

ada = AdaBoostClassifier(
    estimator=base_model,
    n_estimators=200,
    learning_rate=0.5,
    random_state=42
)


# --------------------------------------------------
# 7. Train AdaBoost
# --------------------------------------------------

ada.fit(X_train, y_train)


# --------------------------------------------------
# 8. Make predictions
# --------------------------------------------------

y_pred = ada.predict(X_test)

y_proba = ada.predict_proba(X_test)[:, 1]


# --------------------------------------------------
# 9. Calculate performance
# --------------------------------------------------

accuracy = accuracy_score(y_test, y_pred)

f1 = f1_score(y_test, y_pred)

roc_auc = roc_auc_score(y_test, y_proba)


# --------------------------------------------------
# 10. Display results
# --------------------------------------------------

print("\n==============================")
print("       AdaBoost Results")
print("==============================")

print("Accuracy  :", round(accuracy, 4))
print("F1 Score  :", round(f1, 4))
print("ROC-AUC   :", round(roc_auc, 4))
print("Trees Used:", ada.n_estimators)
```
