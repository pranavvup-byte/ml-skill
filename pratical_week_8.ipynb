```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score
from xgboost import XGBClassifier

# --------------------------------------------------
# 1. Load dataset
# --------------------------------------------------

df = pd.read_csv("/content/sample_data/simple_placement_dataset.csv")

print("Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)


# --------------------------------------------------
# 2. Separate target and features
# --------------------------------------------------

y = df["Placed"]

X = df.drop(columns=["Placed"])


# --------------------------------------------------
# 3. Remove StudentID
# --------------------------------------------------

X = X.drop(columns=["StudentID"])


# --------------------------------------------------
# 4. Split dataset into training and testing
# --------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("\nTraining data:", X_train.shape)
print("Testing data :", X_test.shape)


# --------------------------------------------------
# 5. Create XGBoost model
# --------------------------------------------------

xgb = XGBClassifier(
    n_estimators=1000,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="logloss",
    early_stopping_rounds=30,
    random_state=42,
    n_jobs=-1
)


# --------------------------------------------------
# 6. Train XGBoost
# --------------------------------------------------

xgb.fit(
    X_train,
    y_train,
    eval_set=[(X_test, y_test)],
    verbose=False
)


# --------------------------------------------------
# 7. Make predictions
# --------------------------------------------------

y_pred = xgb.predict(X_test)

y_proba = xgb.predict_proba(X_test)[:, 1]


# --------------------------------------------------
# 8. Calculate performance
# --------------------------------------------------

accuracy = accuracy_score(y_test, y_pred)

f1 = f1_score(y_test, y_pred)

roc_auc = roc_auc_score(y_test, y_proba)


# --------------------------------------------------
# 9. Display results
# --------------------------------------------------

print("\n==============================")
print("        XGBoost Results")
print("==============================")

print("Accuracy  :", round(accuracy, 4))
print("F1 Score  :", round(f1, 4))
print("ROC-AUC   :", round(roc_auc, 4))
print("Trees Used:", xgb.best_iteration + 1)
```
