```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/drive/MyDrive/collab/placement_dataset.csv")

# Placement count by branch
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Branch")
plt.title("Placement Count by Branch")
plt.xlabel("Branch")
plt.ylabel("Count")
plt.xticks(rotation=45)  # helps if branch names are long
plt.tight_layout()
plt.show()

# CGPA distribution
plt.figure(figsize=(6, 4))
plt.hist(df["CGPA"].dropna(), bins=15, color="skyblue", edgecolor="black")
plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Frequency")
plt.show()

# Correlation heatmap
plt.figure(figsize=(6, 5))
numeric_df = df.select_dtypes(include="number")
corr_matrix = numeric_df.corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=True, cmap="viridis", yticklabels=False)
plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.tight_layout()
plt.show()
```
